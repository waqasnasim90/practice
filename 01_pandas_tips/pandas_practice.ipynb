**#16 - Splitting a string into multiple column**

```python
import pandas as pd

df = pd.DataFrame({'name': ['Muhammad Waqas', 'Faisal Waheed', 'Shahbaz Gul', 'Imran Farooq'],
                   'location': ['Karachi, Pakistan', 'Faisalabad,Pakistan', 'Multan,Pakistan', 'Gujrat,Pakistan'],})
df
```

```python
# Spliting a column into two column 
df.name.str.split(' ', expand = True)
```

```python
# adding those splits into a column 
df[["first_name", "last_name"]] = df.name.str.split(' ', expand=True)
df
```

# Location Expand 

```python
# Spliting a column into two column 
df.location.str.split(',', expand = True)
```

```python
# adding those-Location splits into a column 
df[["city_name", "country_name"]] = df.location.str.split(',', expand=True)
df
```

```python
# Refine data manipulation 
df= df [["first_name", "last_name", "city_name", "country_name"]]
df
```
